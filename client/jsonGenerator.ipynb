```python
import pandas as pd
import json
import re

# Load the Excel file
file_path = 'assets/mid-term-exam-schedule_notice-board_sose_243.xlsx'

# Read the Excel file into a pandas DataFrame (assuming the first sheet)
df = pd.read_excel(file_path)

# Define a function to clean up the text
def clean_text(text):
    if pd.notnull(text):  # Check if the text is not NaN
        text = re.sub(r'\s+', ' ', str(text).strip())  # Remove extra spaces and newlines
    return text

# Apply the cleaning function to all string columns in the DataFrame
df = df.apply(lambda col: col.map(clean_text) if col.dtype == "object" else col)

# Convert the cleaned DataFrame to JSON
data_json = df.to_dict(orient='records')

# Save the JSON to a file
output_path = 'output_file.json'
with open(output_path, 'w') as json_file:
    json.dump(data_json, json_file, indent=4)

print(f"JSON file saved to {output_path}")
```

```python
# Reload the Excel file
data = pd.read_excel(file_path)

# Filter data for BSCSE department
bscse_courses = data[data["Dept."] == "BSCSE"]
bscee_courses = data[data["Dept."] == "BSEEE"]
bsds_courses = data[data["Dept."] == "BSDS"]

# Extract unique course titles for BSCSE department
unique_bscse_courses = bscse_courses["Course Title"].dropna().unique()
unique_bscee_courses = bscee_courses["Course Title"].dropna().unique()
unique_bsds_courses = bsds_courses["Course Title"].dropna().unique()
unique_all_courses = data["Course Title"].dropna().unique()


# Split the course titles on '/' and flatten the list
split_bscse_courses = [title.strip() for course in unique_bscse_courses for title in course.split('/')]
split_bscee_courses = [title.strip() for course in unique_bscee_courses for title in course.split('/')]
split_bsds_courses = [title.strip() for course in unique_bsds_courses for title in course.split('/')]
split_all_courses = [title.strip() for course in unique_all_courses for title in course.split('/')]

# Get unique course titles after splitting
unique_split_bscse_courses = list(set(split_bscse_courses))
unique_split_bscee_courses = list(set(split_bscee_courses))
unique_split_bsds_courses = list(set(split_bsds_courses))
unique_split_all_courses = list(set(split_all_courses))

# Sort for better readability
unique_split_bscse_courses.sort()
unique_split_bscee_courses.sort()
unique_split_bsds_courses.sort()
unique_split_all_courses.sort()
unique_split_bscse_courses
```

```python
import json

# Convert the list of unique split BSCSE course titles to JSON format
json_bscse_output = json.dumps(unique_split_bscse_courses, indent=4)
json_bscee_output = json.dumps(unique_split_bscee_courses, indent=4)
json_bsds_output = json.dumps(unique_split_bsds_courses, indent=4)
json_all_output = json.dumps(unique_split_all_courses, indent=4)

# Define the output file path
output_bscse_file_path = "bscse_course_titles.json"
output_bseee_file_path = "bseee_course_titles.json"
output_bsds_file_path = "bsds_course_titles.json"
output_all_file_path = "all_course_titles.json"

# Write the JSON data to a file
with open(output_bscse_file_path, "w") as json_file:
    json_file.write(json_bscse_output)

with open(output_bseee_file_path, "w") as json_file:
    json_file.write(json_bscee_output)

with open(output_bsds_file_path, "w") as json_file:
    json_file.write(json_bsds_output)

with open(output_all_file_path, "w") as json_file:
    json_file.write(json_all_output)

print(f"JSON file saved to {output_bscse_file_path}")
print(f"JSON file saved to {output_bseee_file_path}")
print(f"JSON file saved to {output_bsds_file_path}")
print(f"JSON file saved to {output_all_file_path}")
```
