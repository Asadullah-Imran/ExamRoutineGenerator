```python
import pandas as pd
import json
import re

# Load the Excel file

sose_file_path = 'assets/251_sose.xlsx'
sobe_file_path = 'assets/251_sobe.xlsx'

# Read the Excel file into a pandas DataFrame (assuming the first sheet)
sose_df = pd.read_excel(sose_file_path)
sobe_df = pd.read_excel(sobe_file_path)

# Define a function to clean up the text
def clean_text(text):
    if pd.notnull(text):  # Check if the text is not NaN
        text = re.sub(r'\s+', ' ', str(text).strip())  # Remove extra spaces and newlines
    return text

# Apply the cleaning function to all string columns in the DataFrame
sose_df = sose_df.apply(lambda col: col.map(clean_text) if col.dtype == "object" else col)
sobe_df = sobe_df.apply(lambda col: col.map(clean_text) if col.dtype == "object" else col)

# Convert the cleaned DataFrame to JSON
sose_data_json = sose_df.to_dict(orient='records')
sobe_data_json = sobe_df.to_dict(orient='records')

# Save the JSON to a file
sose_output_path = 'sose_output_file.json'
sobe_output_path = 'sobe_output_file.json'
with open(sose_output_path, 'w') as json_file:
    json.dump(sose_data_json, json_file, indent=4)
with open(sobe_output_path, 'w') as json_file:
    json.dump(sobe_data_json, json_file, indent=4)
print(f"JSON file saved to {sose_output_path}")
print(f"JSON file saved to {sobe_output_path}")

```

```python
# Read the Excel file into a pandas DataFrame (assuming the first sheet)
sose_df = pd.read_excel(sose_file_path)
sobe_df = pd.read_excel(sobe_file_path)


# Extract unique course titles for BSCSE department
sose_unique_all_courses = sose_df["Course Title"].dropna().unique()
sobe_unique_all_courses = sobe_df["Course Title"].dropna().unique()


# Split the course titles on '/' and flatten the list
sose_split_all_courses = [title.strip() for course in sose_unique_all_courses for title in course.split('/')]
sobe_split_all_courses = [title.strip() for course in sobe_unique_all_courses for title in course.split('/')]

# Get unique course titles after splitting
sose_unique_split_all_courses = list(set(sose_split_all_courses))
sobe_unique_split_all_courses = list(set(sobe_split_all_courses))

# Sort for better readability
sose_unique_split_all_courses.sort()
sobe_unique_split_all_courses.sort()
# unique_split_bscse_courses
```

```python
json_sose_course_title = json.dumps(sose_unique_split_all_courses, indent=4)
json_sobe_course_title = json.dumps(sobe_unique_split_all_courses, indent=4)

# Define the output file path
sose_output_file_path = "sose_all_course_titles.json"
sobe_output_file_path = "sobe_all_course_titles.json"



with open(sose_output_file_path, "w") as json_file:
    json_file.write(json_sose_course_title)
with open(sobe_output_file_path, "w") as json_file:
    json_file.write(json_sobe_course_title)


print(f"JSON file saved to {sose_output_file_path}")
print(f"JSON file saved to {sobe_output_file_path}")
```
